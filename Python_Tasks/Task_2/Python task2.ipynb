```python
#Q1. Find the Shortest Word in a Sentence
#Write a Python program to find the shortest word in a given sentence.

#steps 
#1 Find the word with the shortest characters.
#2 Input: A sentence.
#3 Output: The shortest word.
#4 Logic: Split the sentence and compare word lengths.
#5 Start with the first word and update whenever a shorter word appears.
#code

sentence = "Python makes coding easy" 
words = sentence.split()

shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word
  
print(shortest)
```

```python
#Q2. Find the Largest Number in a List
#Write a Python program to find the largest number in a list without using the built-in max() function.
#steps 
#1 Find the largest number.
#2 Input: A list of numbers.[[57,45,44,46,45,35,60,39]]
#3 Output: The largest number.[60]
#4 Logic: Assume the first number is largest. 
#5 Algorithm: Compare every number and update the largest.
#code

numbers = [57,45,44,46,45,35,60,39]
largest = numbers[0]

for num in numbers:
    if num > largest:
     largest = num
print(largest)
```

```python
#Q6. Count Vowels in a String
#Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
#Example input:
#Education
#Expected output:
#5

#steps
#1 Count uppercase and lowercase vowels.
#2 Input: A string.
#3 Output: The vowel count.
#4 Logic: Check whether each character is a vowel.
#5 Algorithm: Start at zero and increase the counter for every vowel.
#code

text = "Education"
vowels = "aeiou"
count = 0

for ch in text.lower():
    if ch in vowels:
        count += 1
print(count)
```

```python
#Q7. Count Consonants in a String
#Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.
#Example input:
#Python 123!
#Expected output:
#5

#steps
#1 Count consonants only.
#2 Input: A string containing letters, digits, and symbols.
#3 Output: The consonant count.
#4 Logic: A consonant is a letter that is not a vowel.
#5 Algorithm: Check that each character is alphabetic and not a vowel.
#code

text = "Python 123!"
vowels = "aeiou"
count  = 0

for ch in text.lower():
    if ch.isalpha() and ch not in vowels:
        count+=1

print(count)
```

```python
#Q8. Find the Most Frequent Character
#Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
#Example input:
#banana
#Expected output:
#a
#1 Find the character with the highest frequenc
#2 Input: A string.
#3 Output: One character.
#4 logic: Count each character, then find the highest frequency while preserving order.
#5 Algorithm and code:
#code

text = "banana"
frequency = {}

for ch in text:
    if ch != " ":
        frequency[ch] = frequency.get(ch, 0) + 1

result = None

for ch in text:
    if ch != " ":
        if result is None or frequency[ch] > frequency[result]:
            result = ch

print(result)
```

```python
#Q9. Find the Least Frequent Character
#Write a Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return the first one encountered. Ignore spaces.
#Example input:
#success
#Expected output:
#u

#steps
#1 Find the character with the lowest frequency.
#2 Input: A string.
#3 Output: One character.
#4 Logic: Count occurrences and find the lowest count.
#5 Algorithm: Scan the characters in order and update when a lower frequency is found.
#code
text = "success"
frequency = {}

for ch in text:
    if ch != " ":
        frequency[ch] = frequency.get(ch, 0) + 1

result = None

for ch in text:
    if ch != " ":
        if result is None or frequency[ch] < frequency[result]:
            result = ch

print(result)
```

```python
#Q10. Find All Characters Appearing More Than Once
#Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.
#Example input:
#programming
#Expected output:
#['r', 'g', 'm']

#1 Find characters appearing more than once.
#2 Input: A string.
#3 Output: A list of repeated characters.
#4 Logic: Count characters and avoid duplicate entries in the result.
#5 Algorithm: Append a character if its frequency exceeds one and it is not already in the result.
#code
text = "programming"
frequency = {}

for ch in text:
    if ch != " ":
        frequency[ch] = frequency.get(ch, 0) + 1

result = []

for ch in text:
    if ch != " ":
        if frequency[ch] > 1 and ch not in result:
            result.append(ch)

print(result)
```

```python
#Q11. Create a Dictionary of Words and Their First Characters
#Write a Python program to create a dictionary in which each word is a key and its first character is the value.
#Example input:
#Python Data Engineering
#Expected output:
#{'Python': 'P', 'Data': 'D', 'Engineering': 'E'
#code
#1 Map each word to its first character.
#2 Input: A sentence.
#3 Output: A dictionary.
#4 Logic: Extract the first character of each word.
#5 Algorithm: Store the word as the key and its first character as the value.

sentence = "Python Data Engineering"
words = sentence.split()
result = {}

for word in words:
    result[word] = word[0]

print(result)
```

```python
#Q12. Create a Dictionary of Words and Their Last Characters
#Write a Python program to create a dictionary in which each word is a key and its last character is the value.
#Example input:
#Python Data Engineering
#Expected output:
#{'Python': 'n', 'Data': 'a', 'Engineering': 'g'}
#1 Map each word to its last character.
#2 Input: A sentence.
#3 Output: A dictionary.
#4 Logic: Extract the last character of every word.
#5 Algorithm: Store each word and its last character.

sentence = "Python Data Engineering"
words = sentence.split()
result = {}

for word in words:
    result[word] = word[-1]

print(result)
```

```python
#Q13. Create a Dictionary of Words and Their Uppercase Versions
#Write a Python program to create a dictionary in which each word is a key and its uppercase version is the value.
#Example input:
#data science python
#Expected output:
#{'data': 'DATA', 'science': 'SCIENCE', 'python': 'PYTHON'}
#1 Convert each word to uppercase.
#2 Input: A sentence.
#3 Output: A dictionary.
#4 Logic: Keep the original word as the key.
#5 Algorithm: Convert each word using upper() for the value.

sentence = "data science python"
words = sentence.split()
result = {}

for word in words:
    result[word] = word.upper()

print(result)
```

```python
#Q14. Create a Dictionary of Numbers and Their Squares
#Write a Python program to create a dictionary containing each number in a list as a key and its square as the value.
#Example input:
#[2, 3, 4, 5]
#Expected output:
#{2: 4, 3: 9, 4: 16, 5: 25}

#1 Associate each number with its square.
#2 Input: A list of numbers.
#3 Output: A dictionary.
#4 Logic: Calculate the square of each number.
#5 Algorithm: Store each number as a key and its square as the value.

numbers = [2, 3, 4, 5]
result = {}

for num in numbers:
    result[num] = num ** 2

print(result)
```

```python
#Q15. Create a Dictionary of Numbers and Their Even/Odd Status
#Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
#Example input:
#[1, 2, 3, 4, 5]
#Expected output:
#{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}
#1 Label every number as even or odd.
#2 Input: A list of integers.
#3 Output: A dictionary.
#4 Logic: Use the remainder after division by two.
#5 Algorithm: Assign "Even" if the remainder is zero; otherwise, assign "Odd".

numbers = [1,2,3,4,5]
result = {}

for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

```python
#Q16. Find the Most Frequent Word in a Sentence
#Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.
#Example input:
#Python is easy and Python is powerful
#Expected output:
#python
#1 Find the most frequent word, ignoring case.
#2 Input: A sentence.
#3 Output: The most frequent lowercase word.
#4 Logic: Normalize the words and count their occurrences.
#5 Algorithm: Build a frequency dictionary and select the word with the highest frequency, preserving the first in a tie.

sentence = "Python is easy and Python is powerful"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = words[0]

for word in words:
    if frequency[word] > frequency[result]:
        result = word

print(result)
```

```python
#Q17. Find All Words Appearing More Than Once
#Write a Python program to find all words that appear more than once in a sentence. Display each repeated word only once, in the order of its first appearance. Treat uppercase and lowercase versions as the same word.
#Example input:
#data science uses data and science tools
#Expected output:
#['data', 'science']
#1 Find every word appearing more than once.
#2 Input: A sentence.
#3 Output: A list of repeated words.
#4 Logic: Convert words to lowercase and count them.
#5 Collect words with frequency greater than one, without duplicates, in their first-appearance order.

sentence = "data science uses data and science tools"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = []

for word in words:
    if frequency[word] > 1 and word not in result:
        result.append(word)

print(result)
```

```python
#Q18. Find the First Word That Appears Only Once
#Write a Python program to find the first word that occurs exactly once in a sentence. Treat uppercase and lowercase versions as the same word.
#Example input:
#apple banana apple mango banana orange
#Expected output:
#mango
#steps
#1 Find the first unique word, ignoring case.
#2 Input: A sentence.
#3 Output: The first word with frequency one.
#4 Logic: Count all lowercase words.
#5 Scan the words in order and select the first whose frequency is one.

sentence = "apple banana apple mango banana orange"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = None

for word in words:
    if frequency[word] == 1:
        result = word
        break

print(result)
```

```python
#Q19. Count Words Whose Length Is Greater Than 4
#Write a Python program to count the words containing more than four characters in a sentence.
#Example input:
#Python makes problem solving interesting
#Expected output:
#4
#1 Count words with more than four characters.
#2 Input: A sentence.
#3 Output: An integer.
#4 Logic: Check the length of each word.
#5 Increase the counter whenever the length exceeds four.

sentence = "Python makes problem solving interesting"
words = sentence.split()
count = 0

for word in words:
    if len(word) > 4:
        count += 1

print(count)
```
